# Kaynak bağlı AYT Tarih QLoRA
Bu notebook iki private Kaggle veri kümesi bekler: `history-tutor-qlora-input-v1` içinde `train.jsonl` ve `dev.jsonl`; `history-tutor-code-v1` içinde bu repodaki QLoRA betiği. Kapalı test, PDF, cevap anahtarı ve öğrenci verisi yüklenmez.

```python
import os
from pathlib import Path
input_root = Path('/kaggle/input')
data_dirs = [path.parent for path in input_root.rglob('train.jsonl') if (path.parent / 'dev.jsonl').is_file()]
runner_files = list(input_root.rglob('history_tutor_qwen14b_qlora.py'))
if len(data_dirs) != 1 or len(runner_files) != 1:
    raise RuntimeError(f'Kaggle girdi paketleri doğrulanamadı: data={data_dirs}, runner={runner_files}')
os.environ['TUTOR_QLORA_INPUT'] = str(data_dirs[0])
os.environ['TUTOR_QLORA_RUNNER'] = str(runner_files[0])
# Bu sayılar, private paket manifestindeki kaggle_runner.environment alanıyla aynı olmalıdır.
os.environ['TUTOR_QLORA_EXPECTED_TRAIN_GROUPS'] = '28'
os.environ['TUTOR_QLORA_EXPECTED_DEV_GROUPS'] = '6'
os.environ['TUTOR_QLORA_SEED'] = '42'
os.environ['TUTOR_QLORA_LR'] = '1e-4'
```

```python
import subprocess, sys
subprocess.run([sys.executable, os.environ['TUTOR_QLORA_RUNNER']], check=True)
# Çıktı: /kaggle/working/history-tutor-qwen3-14b-qlora-v1/report.json ve adapter/
```
